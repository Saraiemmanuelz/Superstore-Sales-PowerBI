```python
### Discount vs Profit

discount_profit = (
    superstore
    .groupby("Discount")["Profit"]
    .agg(["sum", "mean", "count"])
    .reset_index()
)

discount_profit
```

```python
fig = px.scatter(
    discount_profit,
    x="Discount",
    y="sum",
    size="count",
    title="Discount Level and Total Profit",
    labels={
        "Discount": "Discount",
        "sum": "Total Profit ($)",
        "count": "Number of Records"
    },
    hover_data=["mean"]
)

fig.show()
```

### Discount bands and profit margin

Grouping discounts into bands makes the pattern clearer than the scatter above.

```python
superstore["Discount Band"] = pd.cut(
    superstore["Discount"],
    bins=[-0.001, 0, 0.2, 0.4, 1.0],
    labels=["No discount", "Up to 20%", "Over 20% to 40%", "Over 40%"]
)

discount_bands = (
    superstore
    .groupby("Discount Band", observed=True)
    .agg(Lines=("Profit", "count"), Sales=("Sales", "sum"), Profit=("Profit", "sum"))
    .reset_index()
)
discount_bands["Profit Margin %"] = discount_bands["Profit"] / discount_bands["Sales"] * 100

discount_bands.round(1)
```

```python
fig = px.bar(
    discount_bands,
    x="Discount Band",
    y="Profit Margin %",
    title="Profit Margin by Discount Band",
    text_auto=".1f"
)

fig.show()
```

# Superstore EDA — Key Findings and Conclusion

## Key Findings

The exploratory data analysis of the Superstore dataset examined sales,profitability, product categories, sub-categories, customer segments,
regional performance, discounts, and shipping performance.

The dataset generated total sales of $2,297,200.86 and total profit of $286,397.02, with 37,873 units sold. The average discount was 15.62%,
while the average shipping time was 3.96 days.

Technology was the most profitable product category, followed by Office Supplies. Furniture recorded substantially lower total profit compared
with the other major categories.

The West region recorded both the highest sales and highest profit among the four regions. At the customer-segment level, the Consumer segment
recorded both the highest sales and highest profit.

At the sub-category level, Copiers generated the highest profit
($55,617.82), followed by Phones ($44,515.73) and Accessories
($41,936.64). In contrast, Tables recorded the largest loss
(-$17,725.48), followed by Bookcases (-$3,472.56) and Supplies (-$1,189.10).


## Conclusion

The Superstore exploratory data analysis demonstrates differences in sales and profitability across products, regions, and customer segments.
While Technology and the West region contributed strongly to overall performance, some individual sub-categories recorded negative profitability.
The analysis provides useful descriptive insights into the factors and business areas that can be examined further through more detailed
statistical analysis and predictive modelling.
